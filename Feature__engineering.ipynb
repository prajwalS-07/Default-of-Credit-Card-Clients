```python
##It imports the libraries: pandas for tables, numpy for calculations, matplotlib and seaborn for charts, and scipy for statistics.
#It reads cleaned_UCI_Credit_Card.csv into a table called df.
#It renames the long target column default.payment.next.month to default.
#It drops ID, because an ID number has no predictive meaning.
#df.shape prints the size, (29995, 24), which means 29,995 customers and 24 columns.
```

```python
import pandas as pd, numpy as np
import matplotlib.pyplot as plt, seaborn as sns
from scipy import stats

df = pd.read_csv("cleaned_UCI_Credit_Card.csv")
df = df.rename(columns={"default.payment.next.month": "default"}).drop(columns=["ID"], errors="ignore")
print(df.shape)
df.head()
```

```python
##it makes three lists of column names, then builds 9 new columns.
```

```python
bill = [f"BILL_AMT{i}" for i in range(1, 7)]
pay  = [f"PAY_AMT{i}" for i in range(1, 7)]
stat = ["PAY_0", "PAY_2", "PAY_3", "PAY_4", "PAY_5", "PAY_6"]

df["TOTAL_BILL"] = df[bill].sum(axis=1)
df["TOTAL_PAY"]  = df[pay].sum(axis=1)
df["AVG_BILL"]   = df[bill].mean(axis=1)
df["AVG_PAY"]    = df[pay].mean(axis=1)
df["PAY_BILL_RATIO"] = np.where(df["TOTAL_BILL"] > 0, df["TOTAL_PAY"] / df["TOTAL_BILL"], np.nan)
df["PAY_BILL_RATIO"] = df["PAY_BILL_RATIO"].clip(upper=1.5)
df["MAX_DELAY"]      = df[stat].max(axis=1)
df["N_DELAY_MONTHS"] = (df[stat] >= 1).sum(axis=1)
df["UTILIZATION"]    = df["BILL_AMT1"] / df["LIMIT_BAL"]
df["DELAY_TREND"]    = df["PAY_0"] - df["PAY_6"]

new = ["TOTAL_BILL","TOTAL_PAY","AVG_BILL","AVG_PAY","PAY_BILL_RATIO",
       "MAX_DELAY","N_DELAY_MONTHS","UTILIZATION","DELAY_TREND"]
df[new].describe().T
```

```python
##it groups customers into non-defaulters (0) and defaulters (1) and shows the average of each feature for each group
```

```python
df.groupby("default")[new].mean().T.round(2)
```

```python
##it saves the table, now with the 9 new features, as cleaned_with_features.csv (29,995 rows, 33 columns).
```
