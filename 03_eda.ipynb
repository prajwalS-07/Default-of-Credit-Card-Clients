```python
target = 'default.payment.next.month'

print("Target Variable:", target)

print("\nTarget Variable Distribution:")
display(df[target].value_counts())

print("\nTarget Variable Percentage:")
display(df[target].value_counts(normalize=True) * 100)
```

### Explanation

The target variable is
default.payment.next.month.

It indicates whether a customer defaults
on their next month's credit card payment.

- 0 represents non-default.
- 1 represents default.

Studying its distribution helps understand
the proportion of defaulting and
non-defaulting customers.

```python
plt.figure(figsize=(8, 5))

ax = sns.countplot(
    data=df,
    x=target
)

plt.title('Distribution of Credit Card Default')
plt.xlabel('Default Status (0 = No, 1 = Yes)')
plt.ylabel('Number of Customers')

for container in ax.containers:
    ax.bar_label(container)

plt.show()
```

### Observation – Default Distribution

The graph represents the number of customers
who defaulted and those who did not.

The dataset contains a higher proportion of
non-defaulting customers than defaulting
customers.

Approximately 77.88% of customers belong to
the non-default category, while 22.12%
belong to the default category.

This indicates class imbalance, which should
be considered during future machine learning
model development.

```python
plt.figure(figsize=(9, 5))

sns.histplot(
    data=df,
    x='AGE',
    bins=30,
    kde=True
)

plt.title('Age Distribution of Customers')
plt.xlabel('Age')
plt.ylabel('Frequency')

plt.show()
```

### Observation – Age Distribution

The histogram illustrates the distribution
of customer ages.

It helps identify the age groups that are
most represented in the dataset.

The distribution provides an understanding
of the demographic characteristics of
credit card customers.

```python
plt.figure(figsize=(9, 5))

sns.histplot(
    data=df,
    x='LIMIT_BAL',
    bins=40,
    kde=True
)

plt.title('Distribution of Credit Limits')
plt.xlabel('Credit Limit')
plt.ylabel('Frequency')

plt.show()
```

### Observation – Credit Limit Distribution

LIMIT_BAL represents the amount of credit
assigned to each customer.

The histogram helps understand how credit
limits are distributed across customers.

It also helps identify the commonly occurring
credit limit ranges and the variation in
credit allocation.

```python
pay_columns = [
    'PAY_0', 'PAY_2', 'PAY_3',
    'PAY_4', 'PAY_5', 'PAY_6'
]

fig, axes = plt.subplots(2, 3, figsize=(16, 9))

for col, ax in zip(pay_columns, axes.flatten()):
    sns.countplot(data=df, x=col, ax=ax)
    ax.set_title(f'Distribution of {col}')
    ax.set_xlabel('Repayment Status')
    ax.set_ylabel('Customer Count')

plt.tight_layout()
plt.show()
```

### Observation – Repayment Status

The repayment status variables represent
the repayment behaviour of customers
across different periods.

The variables PAY_0, PAY_2, PAY_3, PAY_4,
PAY_5 and PAY_6 help analyse repayment
patterns over time.

Positive values generally indicate payment
delays, while negative and zero values
represent other repayment-status categories.

This analysis helps understand the repayment
behaviour of customers.

```python
bill_columns = [
    'BILL_AMT1', 'BILL_AMT2', 'BILL_AMT3',
    'BILL_AMT4', 'BILL_AMT5', 'BILL_AMT6'
]

fig, axes = plt.subplots(2, 3, figsize=(16, 9))

for col, ax in zip(bill_columns, axes.flatten()):
    sns.histplot(data=df, x=col, bins=30, ax=ax)
    ax.set_title(f'Distribution of {col}')
    ax.set_xlabel('Bill Amount')
    ax.set_ylabel('Frequency')

plt.tight_layout()
plt.show()
```

### Observation – Bill Amount Distribution

BILL_AMT1 to BILL_AMT6 represent the bill
statement amounts across six periods.

The histograms help understand the distribution
and variation of customer bill amounts.

They also help identify highly skewed
distributions and unusual values that may
require further investigation.

```python
payment_columns = [
    'PAY_AMT1', 'PAY_AMT2', 'PAY_AMT3',
    'PAY_AMT4', 'PAY_AMT5', 'PAY_AMT6'
]

fig, axes = plt.subplots(2, 3, figsize=(16, 9))

for col, ax in zip(payment_columns, axes.flatten()):
    sns.histplot(data=df, x=col, bins=30, ax=ax)
    ax.set_title(f'Distribution of {col}')
    ax.set_xlabel('Payment Amount')
    ax.set_ylabel('Frequency')

plt.tight_layout()
plt.show()
```

### Observation – Payment Amount Distribution

PAY_AMT1 to PAY_AMT6 represent the previous
payment amounts made by customers across
six periods.

The histograms help analyse the distribution
of payment amounts and identify variations
in customer payment behaviour.

These variables are important for understanding
how customers repay their credit card bills.

```python
repayment_default = (
    df.groupby('PAY_0')[target].mean() * 100
)

plt.figure(figsize=(9, 5))

repayment_default.plot(kind='bar')

plt.title('Default Rate by Recent Repayment Status')
plt.xlabel('PAY_0 Repayment Status')
plt.ylabel('Default Rate (%)')
plt.xticks(rotation=0)

plt.show()

display(repayment_default)
```

### Observation – Repayment Status vs Default

This analysis examines the relationship
between the most recent repayment status
(PAY_0) and next-month default.

The default rate is calculated for each
repayment-status category.

It helps identify how default rates vary
with different repayment behaviours.

The results show associations and do not
establish causation.

```python
df['credit_limit_group'] = pd.cut(
    df['LIMIT_BAL'],
    bins=5
)

credit_default = (
    df.groupby('credit_limit_group', observed=False)[target]
    .mean() * 100
)

plt.figure(figsize=(10, 5))

credit_default.plot(kind='bar')

plt.title('Default Rate by Credit Limit Group')
plt.xlabel('Credit Limit Group')
plt.ylabel('Default Rate (%)')
plt.xticks(rotation=30)

plt.show()

display(credit_default)
```

### Observation – Credit Limit vs Default

Customers are divided into five credit-limit
groups using the LIMIT_BAL variable.

The default rate is calculated for each group.

This analysis helps examine whether the
proportion of defaulting customers varies
across different credit-limit ranges.

```python
plt.figure(figsize=(9, 5))

sns.boxplot(
    data=df,
    x=target,
    y='PAY_AMT1'
)

plt.title('Payment Amount vs Default Status')
plt.xlabel('Default Status (0 = No, 1 = Yes)')
plt.ylabel('Recent Payment Amount')

plt.show()
```

### Observation – Payment Amount vs Default

The box plot compares recent payment amounts
between defaulting and non-defaulting customers.

It helps compare the median, spread and
distribution of payment amounts across
the two groups.

Potential extreme payment values can also
be identified through this visualisation.

```python
education_default = (
    df.groupby('EDUCATION')[target].mean() * 100
)

plt.figure(figsize=(8, 5))

education_default.plot(kind='bar')

plt.title('Default Rate by Education Category')
plt.xlabel('Education Category')
plt.ylabel('Default Rate (%)')
plt.xticks(rotation=0)

plt.show()

display(education_default)
```

### Observation – Education vs Default

This analysis compares default rates across
different education categories.

It helps understand how default proportions
vary across the categories represented
in the dataset.

The results should be interpreted carefully
and should not be considered evidence
of a causal relationship.

```python
outlier_columns = [
    'LIMIT_BAL',
    'BILL_AMT1',
    'PAY_AMT1',
    'PAY_AMT2'
]

fig, axes = plt.subplots(2, 2, figsize=(14, 9))

for col, ax in zip(outlier_columns, axes.flatten()):
    sns.boxplot(data=df, y=col, ax=ax)
    ax.set_title(f'Outlier Analysis – {col}')

plt.tight_layout()
plt.show()
```

### Observation – Outlier Analysis

Box plots are used to identify potential
outliers in numerical variables.

The box represents the interquartile range
(IQR), while points beyond the whiskers
represent potential extreme observations.

Outliers are not automatically errors.
They may represent genuine customer
financial behaviour.

Therefore, extreme values should be
investigated before deciding whether
they should be retained or removed.

```python
correlation_columns = [
    'LIMIT_BAL',
    'AGE',
    'PAY_0',
    'PAY_2',
    'PAY_3',
    'BILL_AMT1',
    'BILL_AMT2',
    'PAY_AMT1',
    'PAY_AMT2',
    target
]

correlation_matrix = df[
    correlation_columns
].corr()

plt.figure(figsize=(13, 9))

sns.heatmap(
    correlation_matrix,
    annot=True,
    cmap='coolwarm',
    fmt='.2f',
    center=0
)

plt.title('Correlation Heatmap')
plt.tight_layout()
plt.show()
```

### Observation – Correlation Analysis

The correlation heatmap represents the
relationships between selected numerical
variables.

Positive correlation indicates that two
variables tend to increase together, while
negative correlation indicates an inverse
relationship.

The heatmap helps identify variables that
show stronger associations with the target
variable and with each other.

Correlation does not imply causation.

The results can help guide further feature
analysis during machine learning development.

```python
categorical_columns = [
    'SEX',
    'EDUCATION',
    'MARRIAGE'
]

fig, axes = plt.subplots(1, 3, figsize=(16, 5))

for col, ax in zip(categorical_columns, axes):
    sns.countplot(data=df, x=col, ax=ax)
    ax.set_title(f'Distribution of {col}')
    ax.set_xlabel(col)
    ax.set_ylabel('Customer Count')

plt.tight_layout()
plt.show()
```

### Observation – Demographic Analysis

This analysis examines the distribution
of customers across demographic categories.

The variables SEX, EDUCATION and MARRIAGE
provide information about the demographic
composition of the dataset.

The graphs help identify the frequency
of different categories.

```python
print("KEY DATASET FINDINGS")
print("-" * 40)

print("Total customers:", len(df))

print(
    "Non-default customers:",
    (df[target] == 0).sum()
)

print(
    "Default customers:",
    (df[target] == 1).sum()
)

print(
    "Default percentage:",
    round((df[target] == 1).mean() * 100, 2),
    "%"
)

print(
    "Average customer age:",
    round(df['AGE'].mean(), 2)
)

print(
    "Average credit limit:",
    round(df['LIMIT_BAL'].mean(), 2)
)

print(
    "Missing values:",
    df.isnull().sum().sum()
)
```

of the cleaned dataset.

defaulting and non-defaulting customers,
default percentage, average age, average

of the dataset analysed in this notebook.
